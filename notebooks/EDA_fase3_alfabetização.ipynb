```python
# Criando função para analisar estruturas das tabelas
def resumo_dataframe(df, nome):
    print(f"\n{'='*60}")
    print(nome)
    print(f"{'='*60}")

    print(f"Linhas: {df.shape[0]:,}")
    print(f"Colunas: {df.shape[1]}")

    display(pd.DataFrame({
        "tipo": df.dtypes,
        "nulos": df.isna().sum(),
        "% nulos": (df.isna().mean() * 100).round(2),
        "únicos": df.nunique()
    }))
```

```python
resumo_dataframe(alunos, "ALUNOS")
resumo_dataframe(municipio, "MUNICIPIO")
resumo_dataframe(uf, "UF")
resumo_dataframe(meta_municipio, "META MUNICIPIO")
resumo_dataframe(meta_uf, "META UF")
resumo_dataframe(meta_brasil, "META BRASIL")
resumo_dataframe(dicionario_uf, "DICIONARIO UF")
```

- `percentual_participacao`: variável interessante como informação histórica do município. Será avaliada utilizando a participação de 2023 para os alunos de 2024, apesar de não ser esperado um impacto muito grande no modelo.

**Variáveis que não serão utilizadas:**
- `taxa_alfabetizacao`: já temos essa informação em outra base e, após validação, os valores apresentaram o mesmo resultado.
- `meta_alfabetizacao_2025` até `meta_alfabetizacao_2030`: não serão utilizadas, pois representam metas de anos posteriores ao período dos alunos analisados.
- `nivel_alfabetizacao`: não será utilizada, pois é construída a partir do nível de proficiência e, consequentemente, possui relação direta com o target, caracterizando `data leakage`.

```python
# entendenod variavel tp_localização

# Verifica quantos TP_LOCALIZACAO diferentes existem
# para cada combinação município + rede

analise_localizacao = (
    inse
    .groupby(["CO_MUNICIPIO", "TP_TIPO_REDE"])["TP_LOCALIZACAO"]
    .nunique()
    .reset_index(name="qtd_localizacoes")
)

analise_localizacao = analise_localizacao[
    analise_localizacao["qtd_localizacoes"] > 1
]

print(f"Combinações município + rede com mais de uma localização: {len(analise_localizacao)}")
analise_localizacao.head(20)
```

```python
inse[inse['CO_MUNICIPIO'] == 1100049]
```

aqui entendemos que existe mais de um tp_localização para o mesmo municipio e rede, o que nao faz sentido. </br>
pensado nisso, iremos analisar quais sao esses valores duplicados, daremos preferencia para os casos que são MESCLA, vamos identificar quais possuem URBANO e RURAL, sem mescla

```python
analise_localizacao = (
    inse
    .groupby(["CO_MUNICIPIO", "TP_TIPO_REDE"])
    .agg(
        qtd_localizacoes=("TP_LOCALIZACAO", "nunique"),
        tipos_localizacao=("TP_LOCALIZACAO", lambda x: sorted(x.dropna().unique()))
    )
    .reset_index()
)

# Mais de uma localização E nenhuma delas é MESCLA (0)
casos_sem_mescla = analise_localizacao[
    (analise_localizacao["qtd_localizacoes"] > 1) &
    (~analise_localizacao["tipos_localizacao"].apply(lambda x: 0 in x))
]

print(f"Total de municípios + rede com mais de uma localização e sem MESCLA: {len(casos_sem_mescla)}")

casos_sem_mescla.head(40)
```

Aqui vemos que existem 39 casos desses, porém todos são da rede 4 - Privada, que nao existe em nossa tabela de muncipio, ou seja, podemos excluir esses casos e seguir dando preferencia para os casos onde tem mescla

```python
# Identifica quais combinações município + rede possuem MESCLA (0)
chaves_com_mescla = (
    inse.loc[inse["TP_LOCALIZACAO"] == 0, ["CO_MUNICIPIO", "TP_TIPO_REDE"]]
    .drop_duplicates()
)

# Marca as linhas que pertencem a uma chave que possui MESCLA
inse["tem_mescla"] = (
    inse[["CO_MUNICIPIO", "TP_TIPO_REDE"]]
    .merge(
        chaves_com_mescla.assign(tem_mescla=True),
        on=["CO_MUNICIPIO", "TP_TIPO_REDE"],
        how="left"
    )["tem_mescla"]
    .fillna(False)
)

# Se existe MESCLA para município + rede,
# mantém somente a linha MESCLA.
# Se NÃO existe MESCLA, mantém as linhas normalmente.
inse = inse[
    (~inse["tem_mescla"]) |
    (inse["TP_LOCALIZACAO"] == 0)
].copy()

# Remove coluna auxiliar
inse = inse.drop(columns="tem_mescla")
```

```python
# confirmando ajuste

duplicidades = (
    inse
    .groupby(["CO_MUNICIPIO", "TP_TIPO_REDE"])
    .size()
    .reset_index(name="qtd_linhas")
)

duplicidades = duplicidades[duplicidades["qtd_linhas"] > 1]

print(f"Combinações município + rede ainda duplicadas: {len(duplicidades)}")
```
